```python
import pandas as pd
import subprocess
import re
```

```python
df = pd.read_excel("datasets/Top 5 Links x Model.xlsx")
df.head()
```

```python
# List of known brands (lowercased for consistency)
BRANDS = ["ferrari", "lamborghini", "maserati", "bugatti"]

def extract_brand(model: str) -> str:
    model = re.sub(r'[^a-zA-Z0-9]', '', model.lower())
    for brand in BRANDS:
        if brand in model:
            return brand
    return "Unknown"

# Apply to DataFrame
df["brand"] = df["Hashtag"].apply(extract_brand)

# Adding data for files
df['video_path'] = 'data/' + df["brand"] + "/" + df['item_id'].astype(str) + '.mp4'
df['audio_path'] = 'data/' + df["brand"] + "/" + df['item_id'].astype(str) + '.mp3'
filtered_df = df[df["brand"]!="Unknown"]
filtered_df.count()
```

```python
for index, row in filtered_df.iterrows():
    print(row['TikTok Link'])
    subprocess.run(
        [
            "yt-dlp",
            row['TikTok Link'],
            "-o",
            row['video_path'],
            "--cookies-from-browser",
            "chrome"
        ]
    )
```

```python
for index, row in filtered_df.iterrows():
    print(row['audio_path'])
    subprocess.run(
        [
            "ffmpeg",
            "-i",
            row['video_path'],
            "-t",
            "30",
            "-q:a",
            "0",
            "-map",
            "a",
            "-ar",
            "44100",
            "-ac",
            "2",
            row['audio_path']
        ]
    )
```
