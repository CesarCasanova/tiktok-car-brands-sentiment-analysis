```python
import pandas as pd

df = pd.read_csv("data_cleaned.csv")
df.head()
```

```python
from textblob import TextBlob

df['polarity'] = df['translated'].apply(lambda x: TextBlob(str(x)).sentiment.polarity)
df['subjectivity'] = df['translated'].apply(lambda x: TextBlob(str(x)).sentiment.subjectivity)

df['sentiment_label'] = df['polarity'].apply(
    lambda x: 'positive' if x > 0 else 'negative' if x < 0 else 'neutral'
)
# Group by brand for comparison
brand_sentiment_summary = df.groupby('brand')['sentiment_label'].value_counts()
brand_sentiment_summary
```

```python
df.to_csv("./server/sentiment_analysis_textblob.csv")
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()

df['polarity'] = df['translated'].apply(lambda x: analyzer.polarity_scores(x)['compound'])
df['sentiment_label'] = df['polarity'].apply(
    lambda x: 'positive' if x > 0 else 'negative' if x < 0 else 'neutral'
)
# Group by brand for comparison
brand_sentiment_summary = df.groupby('brand')['sentiment_label'].value_counts()
brand_sentiment_summary
```

```python
df.to_csv("./server/sentiment_analysis_vader.csv")
```
